```python
file = open("/kaggle/input/datasets/onkardasari2609/the-complete-works-of-william-shakespeare/pg100.txt",'r')
text = file.read()
```

```python
print('total characters:', len(text))
```

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([text])
```

```python
len(tokenizer.word_index)
```

```python
lines = text.splitlines()
```

```python
sentences = []

for line in lines:
    line = line.strip()

    if line:
        sentences.append(line)
```

```python
print(sentences[0])
print(sentences[1])
print(sentences[2])
```

```python
sequences = []

for sentence in sentences:
    sequence = tokenizer.texts_to_sequences([sentence])[0]
    sequences.append(sequence)

sequences[0]
```

```python
input_sequences = []

for sequence in sequences:
    for i in range(1,len(sequence)):
        input_sequences.append(sequence[:i+1])
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding = 'pre')
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]
```

```python
X
```

```python
y = padded_input_sequences[:,-1]
```

```python
y
```

## We acheived our dataset

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes = 30050)
```

```python
y.shape
```

```python
y
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input,Embedding,LSTM,Dense
```

```python
model = Sequential()

model.add(Input(shape=(56,)))
model.add(Embedding(30050, 100))
model.add(LSTM(150))
model.add(Dense(30050, activation='softmax'))

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

```python
model.fit(X,y,epochs = 10)
```
